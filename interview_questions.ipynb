```python
# Explain shallow copy vs deep copy
import copy
original = {'a':{1,2,3}}
shallow = original.copy

deep = copy.deepcopy(original)

# shallow - Constructs a new outer dict but copies references to the inner nested items
# deep = Recursively clones everything to create an entirely independent copy
```

```python
# For memory visualization
import copy
original = {'a':{1,2,3}}
shallow = original.copy()
deep = copy.deepcopy(original)

print("*** INITIAL MEMORY ADDRESSES ***")
print(f"Outer dict IDs : Original = {id(original)}, Shallow = {id(shallow)}, Deep = {id(deep)}")
print(f"Nested set IDs : Original = {id(original['a'])}, Shallow = {id(shallow['a'])}, Deep = {id(deep['a'])}")

# Mutating the nested set
original['a'].add(4)

print("\n--- AFTER MUTATING NESTED SET")
print(f"Original: {original}")
print(f"Shallow : {shallow}  <-- Affected! Shares the same nested set ID.")
print(f"Deep    : {deep}     <-- Isolated! Has a unique nested set ID.")
```
