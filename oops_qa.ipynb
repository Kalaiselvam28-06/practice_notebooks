## Level 1 – Classes & Objects (Warm-up)

```python
#Q1 - Create a Student class with name and age.
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

s1 = Student("Ravi", 20)
print(f"Name : {s1.name}")
print(f"Age : {s1.age}")
```

```python
# Q2 : Create a Car class with brand and model.
class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model

car1 = Car("Suzuki", "Swift")
print(car1.brand, car1.model)
```

```python
# Q3: Create an Employee class with salary.
class Employee:
    def __init__(self,salary):
        self.salary = salary

employee1 = Employee(20000)
print(employee1.salary)
```

```python
#Q4 : Create a Book class and display details.
class Book:
    def __init__(self, title, author):
        self.title =  title
        self.author = author

    def display_details(self):
        print(f"Title : {self.title}")
        print(f"Author : {self.author}")

book1 =Book("Sun", "Ajay")
book1.display_details()
```

```python
# Q5 :
class Student:
    def __init__(self, name,age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name : {self.name}")
        print(f"Age : {self.age}")

s1 = Student("Ravi", 20)
s2 = Student("Ajay", 21)
s3 = Student("Mohan", 20)

s1.display()
s2.display()
s3.display()
```

```python
class Mobile:
    def __init__(self,brand,model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display(self):
        print(f"Brand : {self.brand}")
        print(f"Model : {self.model}")
        print(f"Price : {self.price}")

mobile1 = Mobile("Samsung", "S25", 50000)
mobile2 = Mobile("Vivo", "abc", 45000)

mobile1.display()
mobile2.display()
```
