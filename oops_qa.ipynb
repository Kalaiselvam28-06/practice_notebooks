```python
# Multiple methods:
class Student_New:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_name(self):
        print(f"Name : {self.name}")

    def display_age(self):
        print(f"Age : {self.age}")

    def display_all(self):
        self.display_name
        self.display_age

s1 = Student_New("Ravi", 26)
s1.display_name()
s1.display_age()
s1.display_all()
```

```python
# One Method Calling Another
class Student_New:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def show_name(self):
        print(f"Name : {self.name}")

    def show_age(self):
        print(f"Age : {self.age}")

    def display(self):
        self.show_name()
        self.show_age()
s1 = Student_New("Ravi", 26)
s1.display()
```

# Level 2 – Methods Performing Operations

```python
# Q1: Rectangle Area
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        rect_area = self.length * self.width
        print(rect_area)

r1 = Rectangle(10,5)
r1.area()
```

```python
# Q2: Circle Circumference
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def circumference(self):
        circ = 2 * 3.14 * self.radius
        print(f"Circumference = {circ}")

circle1 = Circle(7)
circle1.circumference()
```

```python
# Q3 : BankAccount
class BankAccount:
    def __init__(self,name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount <=0:
            print("Deposit not sufficient")
        else:
            self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print('Insufficient funds')

    def display_balance(self):
        return self.balance

acc = BankAccount("Kalai", 1000)
print(f"Initial balance = {acc.balance}")

acc.deposit(500)
acc.withdraw(200)

print(f"After deposit and withdraw, Balance is : {acc.display_balance()}")
```

```python
# Q4 : Calculator
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        result = self.a + self.b
        print(f"Add = {result}")

    def subtract(self):
        result = self.a - self.b
        print(f"Subtract = {result}")

    def multiply(self):
        result = self.a * self.b
        print(f"Multiply = {result}")

    def divide(self):
        result = self.a / self.b
        print(f"Divide = {result}")

cal = Calculator(10,5)
cal.add()
cal.subtract()
cal.multiply()
cal.divide()
```

```python
# Q5: Temperature Converter
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def fahrenheit(self):
        return (self.celsius * 9/5) + 32

t = Temperature(25)
print(t.fahrenheit())
```
