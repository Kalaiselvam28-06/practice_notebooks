```python
#Q4: ATM PIN with Limited Attempts
class ATM:
    def __init__(self, pin):
        self.__pin = pin
        self.__attempts = 0
        self.__blocked = False

    def verify_pin(self, pin):
        if self.__blocked:
            print("Card blocked")
            return
        if self.__pin == pin:
            print("Access Granted")
        else:
            print("Wrong PIN")
            self.__attempts += 1

            if self.__attempts >= 3:
                self.__blocked = True
                print("Card blocked")
pin = ATM(1234)

pin.verify_pin(4444)
pin.verify_pin(1111)
pin.verify_pin(2222)
pin.verify_pin(1234)

```

```python
#Q5 : Temperature with Validation
class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, value):
        if value >= -273.15:
            self.__celsius = value
        else:
            print("Invalid temperature")

    def to_fahrenheit(self):
        return (self.__celsius * 9/5) + 32

t = Temperature(25)

t.set_celsius(30)
t.set_celsius(-500)

print(t.get_celsius())
print(t.to_fahrenheit())
```

```python
#Q6: Bank Account Transaction History
class TransacHistory:
    def __init__(self, balance):
        self.__balance = balance
        self.__transactions = []

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transactions.append(amount)
        else:
            print("Invalid deposit")

    def withdraw(self, amount):
        if amount <=0:
            print("Invalid withdrawal")
        elif amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            self.__transactions.append(-amount)

    def get_balance(self):
        return self.__balance

    def get_transactions(self):
        return self.__transactions

acc = TransacHistory(1000)

acc.deposit(500)
acc.withdraw(200)
acc.deposit(1000)

print(acc.get_balance())
print(acc.get_transactions())
```

```python
#Q8:  Password Manager
class PasswordManager:
    def __init__(self, password):
        self.__password = password

    def change_password(self, old_pass, new_pass):
        if old_pass != self.__password:
            return False
        if len(new_pass) < 8:
            return False
        if new_pass == old_pass:
            return False

        self.__password = new_pass
        return True

    def check_password(self, password):
        if self.__password == password:
            return True
        else:
            return False

pm = PasswordManager("python123")

pm.change_password("wrong", "newpassword")
pm.change_password("python123", "python123")
pm.change_password("python123", "newpass")
pm.change_password("python123", "python45678")

print(pm.check_password("python45678"))
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name):
        self.name =name

    @abstractmethod
    def calculate_salary(self):
        pass

    def display(self):
        print(f"Name : {self.name}")

class FullTimeEmployee(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name)
        self.__salary = salary
        self.bonus = bonus

    def get_salary(self):
        return self.__salary

    def display(self):
        super().display()
        print(f"Monthly salary : ₹{self.get_salary()}")
        print(f"Bonus amount : ₹{self.bonus}")

    def calculate_salary(self):
        cal_salary = self.get_salary()+ self.bonus
        print(f"Total salary : {cal_salary}")

class PartTimeEmployee(Employee):
    def __init__(self, name, hours_worked, hourly_rate):
        super().__init__(name)
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def display(self):
        super().display()
        print(f"Hours worked by {self.name} : {self.hours_worked} hrs")
        print(f"Rate per hour : ₹{self.hourly_rate}")

    def calculate_salary(self):
        cal_salary = self.hours_worked * self.hourly_rate
        print(f"Total salaray : ₹{cal_salary}")

ftime = FullTimeEmployee("Kalai", 50000, 5000)
ftime.display()
ftime.calculate_salary()

print()

parttime = PartTimeEmployee("Ashwin", 6,950)
parttime.display()
parttime.calculate_salary()

#Shorter code
# employees = [ftime,parttime]
# for employee in employees:
#     employee.display()
#     employee.calculate_salary()
```

```python
#Q2:Shopping Cart System
class Product:
    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def get_price(self):
        return self.__price

    def display(self):
        print(f"Product : {self.name}")
        print(f"Price : ₹{self.get_price()}")

class Electronics(Product):
    def final_price(self):
        initial_price = self.get_price()
        discount = initial_price * 0.10
        final_price = initial_price - discount
        return final_price

class Clothing(Product):
       def final_price(self):
            initial_price = self.get_price()
            discount = initial_price * 0.20
            final_price = initial_price - discount
            return final_price

class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_products(self, prodcut):
        self.products.append(prodcut)

    def display_products(self):
        for product in self.products:
            product.display()
            print(f"Final price : ₹{product.final_price()}")
            print()

    def calculate_total(self):
        total = 0

        for product in self.products:
            total += product.final_price()

        return total


laptop = Electronics("Laptop", 50000)
phone = Electronics("Phone", 30000)
shirt = Clothing("Shirt", 2000)

cart = ShoppingCart()

cart.add_products(laptop)
cart.add_products(phone)
cart.add_products(shirt)

cart.display_products()

print(f"Total : ₹{cart.calculate_total()}")
```
