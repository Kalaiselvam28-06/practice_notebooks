```python
#Q1 : Private Balance
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <=0:
            print("Deposit amount must be positive")
        else:
            self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient Balance")
        else:
            self.__balance -= amount

    def set_balance(self, amount):
        if amount <0:
            print("Invalid Balance")
        else:
            self.__balance = amount

acc = BankAccount(1000)
print(acc.get_balance())

#Q2 : Setter Method
acc.set_balance(5000)
print(acc.get_balance())

#Q3 : Deposit
acc.deposit(500)
print(acc.get_balance())

#Q4 : Withdraw
acc.withdraw(1000)
print(acc.get_balance())
```

```python
#Q1: Controlled Bank Account
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid Deposit")
        else:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <0:
            print("Invalid withdrawal")
        elif amount > self.__balance:
            print("Insufficient Funds")
        else:
            self.__balance -= amount

    def display_balance(self):
        return self.__balance

acc1 = BankAccount(5000)
print(f"Initial Balance = {acc1.display_balance()}")

acc1.deposit(2000)
print(f"Balance after deposit = {acc1.display_balance()}")

acc1.withdraw(1500)
print(f"After withdrawal of 1500 = {acc1.display_balance()}")

acc1.withdraw(7000)
print(f"Final balance = {acc1.display_balance()}")

acc1.deposit(-500)
```

```python
# Q2: Employee Salary Protection
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Salary must be positive")
        else:
            self.__salary = salary

    def give_raise(self, amount):
        if amount <= 0:
            print("Raise should be above 0")
        else:
            self.__salary += amount

emp = Employee(30000)

emp.give_raise(5000)
emp.set_salary(-10000)
emp.give_raise(3000)

print(emp.get_salary())
```

```python
#Q3 : Student Marks Validation
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid Marks")

    def add_marks(self, marks):
        if marks <= 0:
            print("Invalid marks")
        elif self.__marks + marks <= 100:
            print("Marks cannot exceed 100")
        else:
            self.__marks += marks

s = Student(70)
s.add_marks(20)
print(f"After adding 20 , the marks will be {s.get_marks()}")
s.add_marks(20)

print(s.get_marks())

    
```

```python
#Q4: ATM PIN with Limited Attempts
class ATM:
    def __init__(self, pin):
        self.__pin = pin
        self.__attempts = 0
        self.__blocked = False

    def verify_pin(self, pin):
        if self.__blocked:
            print("Card blocked")
            return
        if self.__pin == pin:
            print("Access Granted")
        else:
            print("Wrong PIN")
            self.__attempts += 1

            if self.__attempts >= 3:
                self.__blocked = True
                print("Card blocked")
pin = ATM(1234)

pin.verify_pin(4444)
pin.verify_pin(1111)
pin.verify_pin(2222)
pin.verify_pin(1234)

```

```python
#Q5 : Temperature with Validation
class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, value):
        if value >= -273.15:
            self.__celsius = value
        else:
            print("Invalid temperature")

    def to_fahrenheit(self):
        return (self.__celsius * 9/5) + 32

t = Temperature(25)

t.set_celsius(30)
t.set_celsius(-500)

print(t.get_celsius())
print(t.to_fahrenheit())
```

```python
#Q6: Bank Account Transaction History
class TransacHistory:
    def __init__(self, balance):
        self.__balance = balance
        self.__transactions = []

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transactions.append(amount)
        else:
            print("Invalid deposit")

    def withdraw(self, amount):
        if amount <=0:
            print("Invalid withdrawal")
        elif amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            self.__transactions.append(-amount)

    def get_balance(self):
        return self.__balance

    def get_transactions(self):
        return self.__transactions

acc = TransacHistory(1000)

acc.deposit(500)
acc.withdraw(200)
acc.deposit(1000)

print(acc.get_balance())
print(acc.get_transactions())
```

```python
#Q7 : Product Stock Management
class Product:
    def __init__(self,stock):
        self.__stock = stock

    def get_stock(self):
        return self.__stock

    def add_stock(self, quantity):
        if quantity > 0:
            self.__stock += quantity
        else:
            print("Stock must be in n numbers")

    def sell(self, quantity):
        if quantity > self.__stock:
            print("Insufficient stocks")
        else:
            self.__stock -= quantity

p = Product(50)

p.sell(10)
p.add_stock(20)
p.sell(70)
p.sell(30)

print(p.get_stock())
    
```

```python
#Q8:  Password Manager
class PasswordManager:
    def __init__(self, password):
        self.__password = password

    def change_password(self, old_pass, new_pass):
        if old_pass != self.__password:
            return False
        if len(new_pass) < 8:
            return False
        if new_pass == old_pass:
            return False

        self.__password = new_pass
        return True

    def check_password(self, password):
        if self.__password == password:
            return True
        else:
            return False

pm = PasswordManager("python123")

pm.change_password("wrong", "newpassword")
pm.change_password("python123", "python123")
pm.change_password("python123", "newpass")
pm.change_password("python123", "python45678")

print(pm.check_password("python45678"))
```
