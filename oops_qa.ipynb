```python
#Q4: ATM PIN with Limited Attempts
class ATM:
    def __init__(self, pin):
        self.__pin = pin
        self.__attempts = 0
        self.__blocked = False

    def verify_pin(self, pin):
        if self.__blocked:
            print("Card blocked")
            return
        if self.__pin == pin:
            print("Access Granted")
        else:
            print("Wrong PIN")
            self.__attempts += 1

            if self.__attempts >= 3:
                self.__blocked = True
                print("Card blocked")
pin = ATM(1234)

pin.verify_pin(4444)
pin.verify_pin(1111)
pin.verify_pin(2222)
pin.verify_pin(1234)

```

```python
#Q5 : Temperature with Validation
class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, value):
        if value >= -273.15:
            self.__celsius = value
        else:
            print("Invalid temperature")

    def to_fahrenheit(self):
        return (self.__celsius * 9/5) + 32

t = Temperature(25)

t.set_celsius(30)
t.set_celsius(-500)

print(t.get_celsius())
print(t.to_fahrenheit())
```

```python
#Q6: Bank Account Transaction History
class TransacHistory:
    def __init__(self, balance):
        self.__balance = balance
        self.__transactions = []

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transactions.append(amount)
        else:
            print("Invalid deposit")

    def withdraw(self, amount):
        if amount <=0:
            print("Invalid withdrawal")
        elif amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            self.__transactions.append(-amount)

    def get_balance(self):
        return self.__balance

    def get_transactions(self):
        return self.__transactions

acc = TransacHistory(1000)

acc.deposit(500)
acc.withdraw(200)
acc.deposit(1000)

print(acc.get_balance())
print(acc.get_transactions())
```

```python
#Q8:  Password Manager
class PasswordManager:
    def __init__(self, password):
        self.__password = password

    def change_password(self, old_pass, new_pass):
        if old_pass != self.__password:
            return False
        if len(new_pass) < 8:
            return False
        if new_pass == old_pass:
            return False

        self.__password = new_pass
        return True

    def check_password(self, password):
        if self.__password == password:
            return True
        else:
            return False

pm = PasswordManager("python123")

pm.change_password("wrong", "newpassword")
pm.change_password("python123", "python123")
pm.change_password("python123", "newpass")
pm.change_password("python123", "python45678")

print(pm.check_password("python45678"))
```

```python
#Q1: Multilevel Inheritance
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Name : {self.name}")

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name) 
        self.salary = salary

    def show_salary(self):
        print(f"Salary : {self.salary}")

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def show_manager(self):
        self.show_name()
        self.show_salary()
        print(f"Department : {self.department}")

m = Manager("Ravi", 50000, "IT")
m.show_manager()
```

```python
#Q2: 
class Animal:
    def eat(self):
        print("Animal is eating")

class Mammal(Animal):
    def walk(self):
        print("Mammal is walking")

class Dog(Mammal):
    def bark(self):
        print("Dog is barking")

d = Dog()
d.eat()
d.walk()
d.bark()
```

```python
#Q3:
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Cat(Animal):
    def meow(self):
        print("Cat is meowing")

d = Dog()
print("Dog :")
d.eat()
d.bark()

print("\nCat :")
c = Cat()
c.eat()
c.meow()
```

```python
#Q4:
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def display(self):
        print(f"Brand : {self.brand}, Model : {self.model}")

class Bike(Vehicle):
    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine = engine_cc

    def display(self):
        print(f"Brand : {self.brand}, Engine : {self.engine}")

car = Car("Toyota", "Fortunar")
car.display()
bike = Bike("Yamaha", 155)
bike.display()
```

```python
#Q5: Multiple Inheritance
class Father:
    def father_method(self):
        print("Father's method")

class Mother:
    def mother_method(self):
        print("Mother's method")

class Child(Father, Mother):
    def display(self):
        pass

child = Child()
child.father_method()
child.mother_method()
```

```python
#Q6: Multiple Inheritance + Constructors
class Father:
    def __init__(self, name):
        self.name = name

class Mother:
    def __init__(self, age):
        self.age = age

class Child(Father, Mother):
    def __init__(self, name, age, course):
        Father.__init__(self, name)
        Mother.__init__(self, age)
        self.course = course

    def display(self):
        print(f"Name : {self.name}, Age : {self.age}, Course : {self.course}")

c = Child("Kumar", 22, "Python")
c.display()
```

```python
#Q7: Method Overriding + super()
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks")

d = Dog()
d.sound()
```

```python
#Q8: Inheritance + Encapsulation
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <=0:
            raise ValueError("Deposit amount must be above 0")
        self.__balance += amount

class SavingsAccount(BankAccount):
    def display(self):
        print(f"Savings Account Balance : {self.get_balance()}")

acc = SavingsAccount(1000)
acc.deposit(500)
acc.display()
```

```python
#Q9:
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_employee(self):
        print(f"Name : {self.name}, Salary : {self.salary}")

class Developer(Employee):
    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.prolan = programming_language

    def display_developer(self):
        self.display_employee()
        print(f"Programming Language : {self.prolan}")

d = Developer("Arun", 60000, "Python")
d.display_developer()
```

```python
#Q10:
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print(f"Name : {self.name}, Age : {self.age}")

class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course

    def display_student(self):
        self.display_person()
        print(f"Course : {self.course}")

class Employee(Student):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def display_employee(self):
        print(f"Salary : {self.salary}")

class Intern(Student):
    def __init__(self, name, age, course,salary, duration):
        super().__init__(name, age, course)
        self.salary = salary
        self.duration = duration

    def display_intern(self):
        self.display_student()
        print(f"Salary : {self.salary}")
        print(f"Duration : {self.duration}")

intern =Intern("Kumar",22, "Python", 15000, "6 months")
intern.display_intern()
```

```python
#Q8: Diamond Inheritance
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self,name, course):
        super().__init__(name)
        self.course = course

class Employee(Person):
    def __init__(self, name, course, salary):
        super.__init__(name, course)
        self.salary = salary

class Intern(Student, Employee):
    def __init__(self, name,course, salary,duration):
        super().__init__(name, course, salary)
        self.duration = duration

    def display(self):
        print(f"Name : {self.name}")
        print(f"Course : {self.course}")
        print(f"Salary : {self.salary}")
        print(f"Duration : {self.duration}")

intern = Intern("Kumar", "Python", 20000, "6 months")
intern.display()
```
