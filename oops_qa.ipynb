```python
# Q1: Rectangle Area
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        rect_area = self.length * self.width
        print(rect_area)

r1 = Rectangle(10,5)
r1.area()
```

```python
# Q2: Circle Circumference
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def circumference(self):
        circ = 2 * 3.14 * self.radius
        print(f"Circumference = {circ}")

circle1 = Circle(7)
circle1.circumference()
```

```python
# Q3 : BankAccount
class BankAccount:
    def __init__(self,name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount <=0:
            print("Deposit not sufficient")
        else:
            self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print('Insufficient funds')

    def display_balance(self):
        return self.balance

acc = BankAccount("Kalai", 1000)
print(f"Initial balance = {acc.balance}")

acc.deposit(500)
acc.withdraw(200)

print(f"After deposit and withdraw, Balance is : {acc.display_balance()}")
```

```python
# Q4 : Calculator
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        result = self.a + self.b
        print(f"Add = {result}")

    def subtract(self):
        result = self.a - self.b
        print(f"Subtract = {result}")

    def multiply(self):
        result = self.a * self.b
        print(f"Multiply = {result}")

    def divide(self):
        result = self.a / self.b
        print(f"Divide = {result}")

cal = Calculator(10,5)
cal.add()
cal.subtract()
cal.multiply()
cal.divide()
```

## Level 3 – Encapsulation

```python
#Q1 : Private Balance
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <=0:
            print("Deposit amount must be positive")
        else:
            self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient Balance")
        else:
            self.__balance -= amount

    def set_balance(self, amount):
        if amount <0:
            print("Invalid Balance")
        else:
            self.__balance = amount

acc = BankAccount(1000)
print(acc.get_balance())

#Q2 : Setter Method
acc.set_balance(5000)
print(acc.get_balance())

#Q3 : Deposit
acc.deposit(500)
print(acc.get_balance())

#Q4 : Withdraw
acc.withdraw(1000)
print(acc.get_balance())
```

## LEVEL 3 — HARDER ENCAPSULATION PRACTICE

```python
#Q1: Controlled Bank Account
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid Deposit")
        else:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <0:
            print("Invalid withdrawal")
        elif amount > self.__balance:
            print("Insufficient Funds")
        else:
            self.__balance -= amount

    def display_balance(self):
        return self.__balance

acc1 = BankAccount(5000)
print(f"Initial Balance = {acc1.display_balance()}")

acc1.deposit(2000)
print(f"Balance after deposit = {acc1.display_balance()}")

acc1.withdraw(1500)
print(f"After withdrawal of 1500 = {acc1.display_balance()}")

acc1.withdraw(7000)
print(f"Final balance = {acc1.display_balance()}")

acc1.deposit(-500)
```

```python
# Q2: Employee Salary Protection
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Salary must be positive")
        else:
            self.__salary = salary

    def give_raise(self, amount):
        if amount <= 0:
            print("Raise should be above 0")
        else:
            self.__salary += amount

emp = Employee(30000)

emp.give_raise(5000)
emp.set_salary(-10000)
emp.give_raise(3000)

print(emp.get_salary())
```

```python
#Q3 : Student Marks Validation
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid Marks")

    def add_marks(self, marks):
        if marks <= 0:
            print("Invalid marks")
        elif self.__marks + marks <= 100:
            print("Marks cannot exceed 100")
        else:
            self.__marks += marks

s = Student(70)
s.add_marks(20)
print(f"After adding 20 , the marks will be {s.get_marks()}")
s.add_marks(20)

print(s.get_marks())

    
```

```python
#Q4: ATM PIN with Limited Attempts
class ATM:
    def __init__(self, pin):
        self.__pin = pin
        self.__attempts = 0
        self.__blocked = False

    def verify_pin(self, pin):
        if self.__blocked:
            print("Card blocked")
            return
        if self.__pin == pin:
            print("Access Granted")
        else:
            print("Wrong PIN")
            self.__attempts += 1

            if self.__attempts >= 3:
                self.__blocked = True
                print("Card blocked")
pin = ATM(1234)

pin.verify_pin(4444)
pin.verify_pin(1111)
pin.verify_pin(2222)
pin.verify_pin(1234)

```
