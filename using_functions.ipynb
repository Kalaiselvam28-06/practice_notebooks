```python
# Reverse a string
def reverse_string(s):
    return s[::-1].lower()
print(reverse_string("Hello"))
```

```python
# Check if a string is a palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]
print(is_palindrome("Race car"))
```

```python
# Fizzbuzz
def solution(n):
    for i in range(1,n+1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)
solution(15)
```

```python
# Find duplicates in a list
def find_duplicates(lst):
    seen, dupes = set(),set()
    for x in lst:
        if x in seen:
            dupes.add(x)
        seen.add(x)
    # print(seen) - {1,2,3,4}
    # print(dupes) - {1,2}
    # print(list(dupes))
    return list(dupes)

print(find_duplicates([1,2,3,2,4,1]))

# Using list method - Slow
# Why slow : Every time the list code hits if x in seen, seen list scan from beginning to end to check if x exists.
def find_duplicates_list(lst):
    seen, dupes = [], []
    for x in lst:
        if x in seen:
            if x not in dupes:
                dupes.append(x)
        seen.append(x)
    return dupes
print(find_duplicates_list([1,2,3,2,4,1]))
```

```python
# Two sum
def two_sum(nums, target):
    seen = {}
    for i,n in enumerate(nums):
        if target - n in seen:
            return [seen[target-n],i]
        seen[n] = i
    return []

print(two_sum([2,7,11,15],9))
```

```python
# Fibonacci series
def fibonacci(n):
    a,b = 0,1
    result = []
    for i in range(1,n+1):
        result.append(a)
        a,b = b,a+b
    return result
fibonacci(7)
```

```python
# Factorial(Recursive)
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
factorial(5)
```

```python
# Check if two strings are anagrams
def is_anagram(s1,s2):
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()
    return sorted(s1) == sorted(s2)

print(is_anagram("listen", "silent"))
```

```python
# Remove duplicates
def rem_duplicates(lst):
    seen = set()
    result = []
    for x in lst:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result

print(rem_duplicates([1,3,3,2,1,5]))

# If we removed seen
def rem_duplicates(lst):
    # seen = set() - We use seen to remember which numbers already processed
    result = []
    for x in lst:
        result.append(x)
    return result

print(rem_duplicates([1,3,3,2,1,5]))
```

```python
# Count vowels in a string:
def count_vowels(s):
    vowels = "aeiou"
    count = 0
    for ch in s.lower():
        if ch in vowels:
            count += 1
    return count

print(count_vowels("Interview Preparation"))
```

```python
# Find max/min without built-in function
def find_max_min(lst):
    mx = mn = lst[0]
    for x in lst[1:]:
        if x > mx:
            mx = x
        if x < mn:
            mn = x
    return mx, mn

print(find_max_min([4,9,1,7,3]))
```

```python
# Merge two sorted lists
def merge_sorted(a,b):
    result, i, j = [], 0, 0
    while i < len(a) and len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])
    return result

print(merge_sorted([1,3,5],[2,4,6]))
```

```python
# What if lists are unsorted
def merge_any_lists(a,b):
    return sorted(a+b)

list_a = [1,5,3,4]
list_b = [6,2]

final_sorted = merge_any_lists(list_a,list_b)
print(final_sorted)
```

```python
# Flatten a nested list
def flatten(lst):
    result = []
    for item in lst:
        if isinstance(item,list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

print(flatten([1,[2,3,[4,5]],6]))
```

```python
# Find missing number in a array(1 to N)
def find_missing(nums,n):
    total = n * (n+1) // 2
    return total - sum(nums)

print(find_missing([1,2,4,5],5))
```

```python
# Binary search
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high: 
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
print(binary_search([1,3,5,7,9,11], 7))
```

```python
# Bubble sort
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

print(bubble_sort([5,2,9,1,5,6]))
```

```python
# Quick sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + mid + quick_sort(right)

print(quick_sort([9,3,7,1,8,2]))
```

```python
# Find the second largest number in a list
def sec_largest(lst):
    unique = list(set(lst))
    if len(unique) < 2:
        return None
    unique.sort()
    return unique[-2]

print(sec_largest([4,9,1,9,7]))
```

```python
# Find intersection of lists
def intersection(a,b):
    return list(set(a) & set(b))

print(intersection([1,2,3,4],[3,4,5,6]))
```
