```python
# Reverse a string
def reverse_string(s):
    return s[::-1].lower()
print(reverse_string("Hello"))
```

```python
# Check if a string is a palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]
print(is_palindrome("Race car"))
```

```python
# Fizzbuzz
def solution(n):
    for i in range(1,n+1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)
solution(15)
```
