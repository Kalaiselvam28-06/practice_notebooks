```python
# Reverse a string
def reverse_string(s):
    return s[::-1].lower()
print(reverse_string("Hello"))
```

```python
# Check if a string is a palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]
print(is_palindrome("Race car"))
```

```python
# Fizzbuzz
def solution(n):
    for i in range(1,n+1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)
solution(15)
```

```python
# Find duplicates in a list
def find_duplicates(lst):
    seen, dupes = set(),set()
    for x in lst:
        if x in seen:
            dupes.add(x)
        seen.add(x)
    # print(seen) - {1,2,3,4}
    # print(dupes) - {1,2}
    # print(list(dupes))
    return list(dupes)

print(find_duplicates([1,2,3,2,4,1]))

# Using list method - Slow
# Why slow : Every time the list code hits if x in seen, seen list scan from beginning to end to check if x exists.
def find_duplicates_list(lst):
    seen, dupes = [], []
    for x in lst:
        if x in seen:
            if x not in dupes:
                dupes.append(x)
        seen.append(x)
    return dupes
print(find_duplicates_list([1,2,3,2,4,1]))
```

```python
# Two sum
def two_sum(nums, target):
    seen = {}
    for i,n in enumerate(nums):
        if target - n in seen:
            return [seen[target-n],i]
        seen[n] = i
    return []

print(two_sum([2,7,11,15],9))
```

```python
# Fibonacci series
def fibonacci(n):
    a,b = 0,1
    result = []
    for i in range(1,n+1):
        result.append(a)
        a,b = b,a+b
    return result
fibonacci(7)
```
